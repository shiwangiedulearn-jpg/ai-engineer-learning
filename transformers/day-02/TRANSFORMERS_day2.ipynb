```python
from transformers import pipeline
classifier = pipeline("sentiment-analysis")
result = classifier("MediSense is an AI healthcare application.")
print(result)
```

```python
generator = pipeline("text-generation")
result = generator(
    "AI technology",
    max_new_tokens=30
)
print(result[0]["generated_text"])
```
