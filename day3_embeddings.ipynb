```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "heart disease can be associated with a high blood pressure",
    "regular exercise can improve cardiovascular health",
    "diabetes can cause problems with blood sugar regulation",
    "kidney disease can effect the body's ability to remove waste",
    "ml models learn patterns from data",
    "react is a javascript library used to build user interfaces"
]
```

```python
embeddings = model.encode(documents)
print(embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
query = "what can cause cardiovascular problems?"
```

```python
query_embedding = model.encode([query])
```

```python
scores = cosine_similarity(query_embedding, embeddings)[0]
```

```python
results = sorted(
    zip(documents, scores), # it pairs the two things
    key = lambda x:x[1],   #x[0]= sentence and x[1]=score
    reverse= True          # make it in descending order
)
for document, score in results:
  print(round(score,4), document)
```
